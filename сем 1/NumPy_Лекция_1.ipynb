* Атрибуты массивов — определение размера, формы, объема занимаемой памяти и типов данных массивов;

# Атрибуты массивов библиотеки Numpy

```python
import numpy as np
np.random.seed(5) # начальное значение для воспроизводимости последовательности

x1 = np.random.randint(10, size=6) # одномерный массив
x2 = np.random.randint(10, size=(3, 4)) # двумерный массив
x3 = np.random.randint(10, size=(3, 4, 5)) # трехмерный массив

print("массив 1\n", x1, "\n")
print("массив 2\n", x2)
```

```python
print("массив 3\n", x3, "\n")
print("x3 размерность: ", x3.ndim)
print("x3 размер измерений: ", x3.shape)
print("x3 общий размер: ", x3.size)

```

```python
# размер в байтах элементов массива и полный размер массива
print("itemsize:", x3.itemsize, " bytes")
print("nbytes", x3.nbytes, " bytes")
```

# Срезы массивов (:)

Общее правило так же как и в циклах:

x[начало:конец:шаг]

Если не указано, то по умолчанию шаг - 1, начало - 0, конец - размер измерения

Одномерные подмассивы

```python
x = np.arange(10)
x
```

```python
# первые 5
print(x[:5])

# все после индекса 5
print(x[5:])

# подмассив из середины
print(x[4:8])
```

```python
# каждый второй
print(x[::2])

# каждый третий
print(x[::3])

# каждый второй, начиная с 3-его индекса
print(x[3::2])
```

Обратите особое внимание когда меняете шаг на отрицательный

(В таком случае значения для начало и конец меняются местами)

```python
# пример перевернуть массив
print(x[::-1])

# Каждый второй в обратном порядке начиная с индекса 6
# С 6-ого индекса по 0-ой с шагом -2
print("Исходный - ", x)
print(x[6::-2])
```

# Изменение формы массивов

Например, если у нас есть список чисел, а мы хотим его разместить в двумерной матрице, мы можем воспользоваться функцией reshape( (размер, размер) )

```python
grid = np.arange(1, 10).reshape((3, 3))
print(grid)
```

**Размер исходного массива должен соответствовать размеру измененного.**

Преобразование одномерного массива в двумерную матрицу-строку или матрицу-столбец (можно как методом reshape или специальным методом newaxis)

```python
x = np.array([1, 2, 3])

# Преобразование в вектор-строку с reshape
x.reshape((1, 3))
```

```python
# Преобразование в вектор-строку с newaxis
x[np.newaxis, :]
```

```python
# Преобразование в вектор-столбец с помощью reshape
x.reshape((3, 1))
```

```python
# Преобразование в вектор-столбец с помощью newaxis
x[:, np.newaxis]
```

# Слияние и разбиение массивов

Методы np.concatenate, np.vstack, np.hstack

Concatenate

```python
x = np.array([1, 2, 3])
y = np.array([3, 2, 1])
np.concatenate([x, y])
```

```python
# Объединение нескольких одновременно
z = [99, 99, 99]
print(np.concatenate([x, y, z]))
```

```python
# Объединение двумерных массивов
grid = np.array([[1, 2, 3],
                 [4, 5, 6]])

# Слияние происходит по первой оси координат
np.concatenate([grid, grid])
```

```python
# Слияние по второй оси координат (индекс 0)
np.concatenate([grid, grid], axis=1)
```

concatenate - хорошо использзовать для работы с одинаковыми измерениями, если измерения отличаются, то...

```python
x = np.array([1, 2, 3])
grid = np.array([[9, 8, 7],
                [6, 5, 4]])
```

```python
# Объединяет массивы по вертикали
np.vstack([x, grid])
```

```python
# Объединяет массивы по горизонтали
y = np.array([[99],
              [99]])
np.hstack([grid, y])
```

# Арифметические функции

```python
x = np.arange(4)
print("x =", x)
print("x + 5 =", x + 5)
print("x - 5 =", x - 5)
print("x * 2 =", x * 2)
print("x / 2 =", x / 2)
print("x // 2 =", x // 2) # целочисленное деление
print("-x = ", -x) # изменение знака
print("x ** 2 = ", x ** 2) # возведение в степень
print("x % 2 = ", x % 2) # деление по модулю
```

# Специализированные универсальные функции

```python
from scipy import special

# Гамма-функции (обобщенные факториалы) и тому подобные функции
x = [1, 5, 10]
print("gamma(x) =", special.gamma(x))
print("ln|gamma(x)| =", special.gammaln(x))
print("beta(x, 2) =", special.beta(x, 2))
```

# Булевы операторы

нужна информация обо
всех днях с толщиной слоя осадков менее четырех дюймов и более одного дюйма

```python
np.sum((inches > 0.5) & (inches < 1))
```

```python
print("Количество дней без дождя: ", np.sum(inches == 0))
print("Количество дней с дождем: ", np.sum(inches != 0))
print("Дни с уровнем осадков > 0.5 дюйма:", np.sum(inches > 0.5))
print("Дни с уровнем осадков < 0.2 дюйма::", np.sum((inches > 0) &
(inches < 0.2)))
```
