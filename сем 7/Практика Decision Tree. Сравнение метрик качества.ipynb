## **4. Оценка и сравнение метрик качества**

Мы рассмотрим следующие метрики:
*   **Accuracy (Точность)**: Доля правильных ответов модели.
*   **Precision (Точность)**: Способность модели не называть положительный класс отрицательным. `TP / (TP + FP)`
*   **Recall (Полнота)**: Способность модели находить все положительные примеры. `TP / (TP + FN)`
*   **F1-score**: Гармоническое среднее Precision и Recall. Хорошо работает при дисбалансе классов.
