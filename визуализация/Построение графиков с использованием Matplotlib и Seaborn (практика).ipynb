# №1 Случайные блуждания и тренды

Задача моделирует случайное перемещение на n точек по плоскости, можно представить, что это модель случайного передвижения муравья.

```python
import numpy as np
import matplotlib.pyplot as plt
from random import choice

class RandomWalk:
    """Класс для генерации случайных блужданий."""

    def __init__(self, num_points=5000):
        """Инициализация атрибутов блуждания."""
        self.num_points = num_points
        self.x_values = [0]
        self.y_values = [0]

    def fill_walk(self):
        """Вычисление всех точек в блуждании."""
        while len(self.x_values) < self.num_points:
            # Решения о направлении и расстоянии
            x_direction = choice([1, -1])
            x_distance = choice([0, 1, 2, 3, 4])
            x_step = x_direction * x_distance

            y_direction = choice([1, -1])
            y_distance = choice([0, 1, 2, 3, 4])
            y_step = y_direction * y_distance

            # Отбрасываем движения, которые никуда не ведут
            if x_step == 0 and y_step == 0:
                continue

            # Вычисляем новые значения
            next_x = self.x_values[-1] + x_step
            next_y = self.y_values[-1] + y_step

            self.x_values.append(next_x)
            self.y_values.append(next_y)
```
