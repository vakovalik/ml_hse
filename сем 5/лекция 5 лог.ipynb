# Логистическая регрессия

это модель машинного обучения используемая для бинарной классификации.

Классификация - разделение данных по какому-либо признаку.

Классификация бывает:

1.   бинарная - классов всего 2 (яблоки/груши, здоровый/больной, спам/не спам)
2.   мультиклассовая или полиномиальная классификация - 3 и более классов (кошка, собака, кролик, лошадь)

**Название**

регрессия - получение взаимосвязи между независимой и зависимой переменной

логистическая - по названию [логистической функции](https://ru.wikipedia.org/wiki/%D0%9B%D0%BE%D0%B3%D0%B8%D1%81%D1%82%D0%B8%D1%87%D0%B5%D1%81%D0%BA%D0%BE%D0%B5_%D1%83%D1%80%D0%B0%D0%B2%D0%BD%D0%B5%D0%BD%D0%B8%D0%B5), которая в ней используется

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-10, 10, 100)
u = np.random.random(100)
y = 0.5 * x + u > 0
plt.scatter(x, y)
```

Данные по оси X могут быть самые разные, главное, чтобы они были сгруппированы, а по у существует только 2 вида значений (2 класса)
 - вес человека в килограммах - оценка ожирения
 - размер опухоли в миллиметрах - злокачественная или нет
 - доход на семью - желание купить машину

**Гипотеза**

$h = g(θx)$

$g(z) = \frac {1}{1+e^{-z}}$

$h = \frac {1}{1+e^{-θx}}$


```python
xs = np.linspace(-10, 10, 100)
ys = 1/(1 + np.exp(-xs))
plt.plot(xs, ys)
plt.axhline(y=0.5, color='g', linestyle='-')
```

Смысл этой функции в преобразовании чисел

 - если по x был 0 - она его преобразует в 0.5
 - любое большое положительное число преобразуется в 1
 - любое маленькое отрицательное число преобразуется в 0

  



```python
plt.axhline(y=0.5, color='g', linestyle='-')
plt.scatter(x, y)
plt.plot(xs, ys, color = 'red')
```

**Принципиальная разница линейной и логистической регрессии**

линейная регрессия - h показывает реальное числовое значение
логистическая регрессия - h показывает вероятность отнесения объекта к тому или иному классу

h < 0.5 - данные относятся к классу 0

h > 0.5 - данные относятся к классу 1

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression()
model.fit(x.reshape(-1,1), y.reshape(-1, 1))
h = model.predict(x.reshape(-1,1))

plt.scatter(x,y)
plt.plot(x,h, color = 'r')
```

**Рассчитаем работу простой сети**

Снимок экрана 2021-11-11 в 23.38.39.png

```python
import numpy as np

x1 = 1
x2 = 0.5
w11 = 0.9
w12 = 0.2
w21 = 0.3
w22 = 0.8
sum1 = x1 * w11 + x2 * w21
sum2 = x1 * w12 + x2 * w22
out1 = 1/(1 + np.exp(-sum1))
out2 = 1/(1 + np.exp(-sum2))
print("Нейрон 1  сумма: %f.2  выход: %f.2" % (sum1, out1))
print("Нейрон 2  сумма: %f.2  выход: %f.2" % (sum2, out2))
```

```python
import matplotlib.pyplot as plt
xs = np.linspace(-10, 10, 100)
ys = 1/(1 + np.exp(-xs))

fig, ax = plt.subplots()
ax.plot(xs, ys)
ax.grid()

#  Вертикальные линии:
ax.vlines(0, ys.min(), ys.max()) # ось y
ax.vlines(sum1, 0, out1, color = 'r')
ax.vlines(sum2, 0, out2, color = 'g')

#  Горизонтальные динии:
ax.hlines(0, xs.min(), xs.max()) # ось x
ax.hlines(out1, 0, out1, color = 'r')
ax.hlines(out2, 0, out2, color = 'g')
```

**Зачем нужна функция активации?**

1. Математическое объяснение. ФА отделяет один слой нейронной сети (классификатор) от другого. Если бы его не было, то разделять сеть на слои не было бы смысла т.к. можно было бы объединить сумму весов нейронов первого слоя с суммой весов нейронов последующих слоев и все они вместе реботали бы как один нейрон.

2. Интуитивное объяснение.

image.png

**Как понять сколько нейронов на каком слое использовать?**

**Входной слой**: количество нейронов = количеству входных переменных

Пример 1: если у нас есть набор данных с 10 колонками, и все 10 колонок мы хотим использовать для обучения - то входной слой будет состоять из 10 нейронов

Пример 2: если мы на вход подаем картинку, имеющую разрешение 10 на 20 пикселей, то количество входных нейронов = 10 * 20 = 200

**Выходной слой**: количество нейронов = количеству классов, которые мы хотим распознать

**Скрытые слои**: количество слоёв и количество нейронов на каждом слое - однозначного ответа нет. Для простых задач обычно хватает одного скрытого слоя. В классическом машинном обучении больше 3 скрытых слоев обычно не используют. Подбирают количество слоев и количество нейронов в каждом слое методом проб и ошибок ввиду сложности интерпритации что происходит внутри. В глубоком обучении появились разные архитектуры сетей и разные типы слоев, несущие свой смысл, поэтому подбирать количество слоев и нейронов стало проще.
